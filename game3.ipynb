음료수를 구매 할 수 있는 자판기 입니다. 음료수 종류 및 개당 금액
콜라 : 500
사이다 : 450
맥콜 300

```python
#강사님 문제 완료

콜라 = 500
사이다 = 450
맥콜 = 300
선택한음료 = input("콜라, 사이다, 맥콜 중 하나를 주문하세요")
금액 = 0
if 선택한음료 == '콜라':
    print('콜라를 선택하셨습니다. 500원입니다.')
    금액 = 500
elif 선택한음료 == '사이다':
    print('사이다를 선택하셨습니다. 450원입니다.')
    금액 = 450
elif 선택한음료 == '맥콜':
    print('매콜를 선택하셨습니다. 300원입니다.')
    금액 = 300
else :
    print('잘못 입력하셨습니다.')
    
수량 = int(input('음료수 수량을 선택해주세요'))
음료값 = 금액 * 수량
print(f'{선택한음료} {수량}개 금액은{음료값}원 입니다.')

지불금액=int(input('지불할금액을 입력하세요'))
거스름돈 = 지불금액 - 음료값
print(f'{지불금액}원을 받고 {거스름돈}원 거스름돈 받으세요')
```

```python
#선택이 잘못 되거나 입력이 잘못되었을 때

콜라 = 500
사이다 = 450
맥콜 = 300
선택한음료 = input("콜라, 사이다, 맥콜 중 하나를 주문하세요")
금액 = 0
if 선택한음료 == '콜라':
    print('콜라를 선택하셨습니다. 500원입니다.')
    금액 = 500
elif 선택한음료 == '사이다':
    print('사이다를 선택하셨습니다. 450원입니다.')
    금액 = 450
elif 선택한음료 == '맥콜':
    print('매콜를 선택하셨습니다. 300원입니다.')
    금액 = 300
else :
    print('잘못 입력하셨습니다.')
    
수량 = int(input('음료수 수량을 선택해주세요'))
음료값 = 금액 * 수량
print(f'{선택한음료} {수량}개 금액은{음료값}원 입니다.')

지불금액=int(input('지불할금액을 입력하세요'))
거스름돈 = 지불금액 - 음료값
if 지불금액 >= 음료값 :
    print(f'{지불금액}원을 받고 {거스름돈}원 거스름돈 받으세요')
elif 지불금액 <= 음료값 :
    print(f'{지불금액}원을 받았는데 {거스름돈}원이 부족합니다.')
    지불금액=int(input('지불할금액을 다시입력하세요'))
else :
    print(f'{지불금액}원을 받고 {거스름돈}원 입니다. 감사합니다')
```

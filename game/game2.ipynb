문제) 간단한 계산기를 만드세요

```python
def 더하기(a,b):
    return a+b
```

```python
def 빼기(a,b):
    return a-b
```

```python
def 곱하기(a,b):
    return a*b
```

```python
def 나누기(a,b):
    return a/b
```

```python
print("간단한 계산기 입니다.")
print(f"+:{더하기(10,1)}")
print(f"-:{빼기(10,1)}")
print(f"*:{곱하기(10,2)}")
print(f"/:{나누기(10,5)}")
```

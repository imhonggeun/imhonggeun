문자열 표현하는 방법
1.",'하나로 사용했을 경우는 한줄
2.",'세개로 사용했을 경우는 여러줄로 사용 가능

```python
a = "Hello World"
b = 'Python is fun'
c = """Life is 
too short, You need 
python"""
d = '''Life is too short, 
   You need python'''
print(a, b, c, d)
```

```python
a = '"Python" is fun'
print(a)
```

```python
'''
1. a 변수에 정수의 값을 넣기
2. a 변수를 콘솔에 값을 출력하기
'''
a = 100
print(a)
```

```python
a = 'Life is too short \n You need \t python'
print(a)
```

```python
a = 'Life is too short You need python'
# print(a + '\n' * 3) 이건 \n을 3번 곱하기 때문에  a를 한줄나오고 \n이 새번나온다
print((a + '\n') * 3) #이건 (a + '\n')을 3번 곱하기 때문에 내용이 3번 나온다
```

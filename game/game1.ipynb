```python
데이터 ={"k1" : "값1"} #dict
print(type(데이터))
#print(데이터)
```

```python
print(데이터["k1"]) #키값을 통해 값을 가지오는 것
```

```python
데이터["k1"] = 10 # 키 k1에 10을 담는 것
print(데이터)#print(데이터["k1"]) => 이렇게 하면 10만나옴
```

```python
데이터[2] = 100
print(데이터)
```

```python
데이터["k3"] =[] #데이터 딕션에 k3를 추가하고 리스트를 담는것
데이터["k4"] =() #데이터 딕션에 k4를 추가하고 튜플를 담는것
print(데이터)
```

```python
데이터[True] = 1 #True 라는 키값을 추가
print(데이터)
```

```python
for i in 데이터: #데이터 딕션을 i라는 변수에 담는다
    print(데이터[i]) #담은 i변수를 출력하면 값을 가지고온다
```

```python
print("사용자의 이름과 나이를 입력 받아 출력하기")
사용자목록 = [] #리스트

while True: #
    사용자 = {}
    name = input("당신의 이름을 입력해주세요.")
    if name == '': 
        break #입력값이 없을 때 멈춤
    사용자 ['name'] = name

    age = input("당신의 나이를 입력해주세요.")
    사용자['age'] = age 
    
    사용자목록.append(사용자) #이름과 나이를 담아서 사용자목록에 넣는다
print(사용자목록)

while True:
    선택 = input("찾으실 사용자 이름을 입력하세요") # 위에 저장한 이름을 찾기
    if 선택=='':
        break
    for r in 사용자목록:
        if 선택 == r['name']: # 위에 값에 저장한 이름과 동일하면 이름과 나이를 출력
            print(f'{r['name']}는 {r['age']}살 입니다.')
```

```python
단 = range(1,10)
for c in 단 :
     print(c)
     for d in 단 :
        print(f"{c} * {d} = {c*d}")
```

```python
결과 = [1]
for 단 in range(1,10):
    빈배열 = []
    for 곱 in range(1,10):
        빈배열.append(f"{단} * {곱} = {단 * 곱}")
    결과.append(빈배열)
print(결과)
```

```python
곱단 = range(1,10)
결과 = []
for 단 in 곱단:
    빈배열 = []
    for 곱 in 곱단:
        빈배열.append(f"{단} * {곱} = {단 * 곱}")
    결과.append(빈배열)
print(결과)
```

```python
import sys # 관리자 모드에서 프로그램을 바로 종료하기 위해 import

# 자판기 상품 딕셔너리
products = {
    '1': {'name': '콜라', 'price': 1000, 'stock': 5},
    '2': {'name': '사이다', 'price': 1200, 'stock': 3},
    '3': {'name': '물', 'price': 800, 'stock': 10},
    '4': {'name': '오렌지주스', 'price': 1500, 'stock': 2}
}

# 투입된 돈
money_inserted = 0
# 총 매출
total_sales = 0

# 관리자 모드 함수
def admin_mode():
    global total_sales
    while True:
        print("\n===== 관리자 모드 =====")
        print("1. 재고 추가")
        print("2. 매출 확인")
        print("3. 사용자 모드로 돌아가기")
        print("4. 시스템 종료")
        admin_choice = input("작업을 선택하세요: ")

        if admin_choice == '1':
            p_num = input("재고를 추가할 음료 번호를 입력하세요: ")
            if p_num in products:
                try:
                    add_stock = int(input(f"'{products[p_num]['name']}'에 추가할 수량을 입력하세요: "))
                    if add_stock > 0:
                        products[p_num]['stock'] += add_stock
                        print(f"재고가 성공적으로 추가되었습니다. 현재 재고: {products[p_num]['stock']}개")
                    else:
                        print("추가할 수량은 0보다 커야 합니다.")
                except ValueError:
                    print("숫자로 입력해주세요.")
            else:
                print("존재하지 않는 상품 번호입니다.")

        elif admin_choice == '2':
            print(f"현재까지의 총 매출은 {total_sales}원 입니다.")

        elif admin_choice == '3':
            print("사용자 모드로 돌아갑니다.")
            break # 관리자 모드 루프 탈출

        elif admin_choice == '4':
            print("시스템을 종료합니다.")
            sys.exit() # 프로그램 완전 종료

        else:
            print("잘못된 선택입니다.")


# 메인 루프
while True:
    print("\n===== 미니 자판기 =====")
    for num, info in products.items():
        status = "[품절]" if info['stock'] == 0 else f"[재고: {info['stock']}개]"
        print(f"{num}. {info['name']} ({info['price']}원) {status}")

    print("=======================")
    print(f"현재 투입된 금액: {money_inserted}원")
    print("-----------------------")
    print("1. 돈 넣기")
    print("2. 음료 선택")
    print("3. 잔돈 반환 및 종료")
    print("-----------------------")

    try:
        choice = input("원하는 작업을 선택하세요 (관리자 모드는 'admin' 입력): ")

        if choice == 'admin':
            admin_mode()
            continue # 관리자 모드 종료 후 메인 메뉴 다시 표시

        # 1. 돈 넣기
        if choice == '1':
            amount_to_add = int(input("넣을 금액을 입력하세요: "))
            if amount_to_add > 0:
                money_inserted += amount_to_add
                print(f"{amount_to_add}원이 투입되었습니다.")
            else:
                print("금액은 양수로 입력해야 합니다.")

        # 2. 음료 선택
        elif choice == '2':
            product_num = input("원하는 음료의 번호를 입력하세요: ")

            if product_num in products:
                product = products[product_num]
                if money_inserted >= product['price']:
                    if product['stock'] > 0:
                        money_inserted -= product['price']
                        product['stock'] -= 1
                        total_sales += product['price']  # 매출 기록
                        print(f"\n>> {product['name']}이/가 나왔습니다! <<")
                        print(f"남은 금액: {money_inserted}원")
                    else:
                        print("죄송합니다. 해당 음료는 품절되었습니다.")
                else:
                    print("금액이 부족합니다. 돈을 더 넣어주세요.")
            else:
                print("올바른 상품 번호를 입력해주세요.")

        # 3. 잔돈 반환 및 종료
        elif choice == '3':
            if money_inserted > 0:
                print(f"잔돈 {money_inserted}원이 반환됩니다.")
            print("자판기 이용을 종료합니다. 감사합니다.")
            break

        else:
            print("잘못된 선택입니다. 1, 2, 3 또는 'admin'을 입력해주세요.")

    except ValueError:
        print("유효한 값을 입력해야 합니다. 다시 시도해주세요.")
    except Exception as e:
        print(f"알 수 없는 오류가 발생했습니다: {e}")
```

2. 슬라이싱 : 변수[시작위치값:끝위치값]

```python
a = "Likf is too short, You need Python"
print(a[8]+a[9]+a[10])
```

```python
a = "Likf is too short, You need Python"
print(a[8:11]) #슬라이싱으로 변수 끝위차값을 나타날 땐 -1이 된다고 생각하면된다 
               #too 가 8:10 이지만 8:11 되는이유
```

```python
a = "Life is too short, You need Python"
print(a[-7:]) # 끝자리 값을 주지 않으면 마지막을 뜻하기 때문에 Python까지 다 볼 수 있다.
```

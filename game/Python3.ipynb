```python
a = "Python"
# need 이후에 띄어쓰기 적용,%다음에 띄어쓰기 상관없이 적용
# 문자같은 경우 String(s) 로 표현해서 적용
print("Life is too short, You need %s" %a)
```

```python
a = 100
# 숫자 같은 경우 decimal(d)로 해서 적용
# s로 해도 된다 문자열안에 숫자가 포함되어 있기 때문에
print("Life is too short, You need %d"%a)
```

```python
a ="Python"
#format(f)를 사용해서 좀 더 가독성이 높고 편하게 적용 수 있고 숫자문자 상관없다
print(f"Life is too short, You need {a}")
```

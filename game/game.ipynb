```python
시작 = 1
끝 = 10
while 시작 < 끝 :
    print(f"출력 : {시작}")
    
    곱=1
    while 곱 < 끝 :
        print(f"{시작} * {곱} = {시작 * 곱}")
        곱 = 곱 +1
    시작 = 시작 + 1
```

```python
while 선택한음료 :
    선택한음료 = input("콜라, 사이다, 맥콜 중 하나를 주문하세요")
    금액 = 0
    if 선택한음료 == '콜라':
        print('콜라를 선택하셨습니다. 500원입니다.')
        금액 = 500
    elif 선택한음료 == '사이다':
        print('사이다를 선택하셨습니다. 450원입니다.')
        금액 = 450
    elif 선택한음료 == '맥콜':
        print('매콜를 선택하셨습니다. 300원입니다.')
        금액 = 300
    else :
        print('잘못 입력하셨습니다.')
        
    수량 = int(input('음료수 수량을 선택해주세요'))
    음료값 = 금액 * 수량
    print(f'{선택한음료} {수량}개 금액은{음료값}원 입니다.')

    지불금액=int(input('지불할금액을 입력하세요'))
    거스름돈 = 지불금액 - 음료값
    if 지불금액 >= 음료값 :
        print(f'{지불금액}원을 받고 {거스름돈}원 거스름돈 받으세요')
    elif 지불금액 <= 음료값 :
        print(f'{지불금액}원을 받았는데 {거스름돈}원이 부족합니다.')
        지불금액=int(input('지불할금액을 다시입력하세요'))
    else :
        
        print(f'{지불금액}원을 받고 {거스름돈}원 입니다. 감사합니다')
```

```python
시작 = 1
끝 = 100
while 시작 < 끝:
    선택한음료 = input('버튼을 눌러 음료수를 선택해 주세요: ')
    if 선택한음료 == '콜라' or 선택한음료 == '칠성사이다' or 선택한음료 == '환타' or 선택한음료 == '레몬녹차':
       
        if 선택한음료 == '콜라':
            금액 = 1700
            print('음료 선택 콜라 1700원')
           
        elif 선택한음료 == '칠성사이다':
            금액 = 1500
            print('음료 선택 칠성사이다 1500원')
           
        elif 선택한음료 == '환타':
            금액 = 1600
            print('음료 선택 환타 1600원')
           
        elif 선택한음료 == '레몬녹차':
            금액 = 1000
            print('음료 선택 레몬녹차 1000원')
           
        수량 = int(input('수량을 입력하세요: '))
       
        총금액 = 금액 * 수량
        print(f'{선택한음료} {수량}개 금액은 {총금액}원 입니다.')


        지불금액 = int(input('지불할 금액을 입력하세요: '))
        잔돈 = 지불금액 - 총금액
        print(f'{지불금액}원을 받고 {잔돈}원을 잔돈으로 뽑아주세요.')


    else:
        print('다른 음료는 판매하지 않습니다.')



```

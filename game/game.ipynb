반복문에서 사용되는 기능

break : 강제로 종료
continue : 다음으로 점프

1.break 사용 예시

```python
변수 = 0
while True:
    변수 +=1
    if 변수 == 5:
        break
    else:
        print(변수)
```

2.continue 사용 예시

```python
변수 = 0
while 변수 <5:
    변수 +=1
    if 변수%2 ==0:
        continue
    else:
        print(변수)
```

```python
a = True
b = 0
while a:
    if b== 0:
        print(b)
    b +=1
    if b==4:
        continue
    print(b)
```

```python
a = True
b = 0
while a:
    
    if b==4:
        b +=1
        continue
    print(b)
    b +=1
```

```python
시작 =1 
끝 =10

while 시작 < 끝:
    if 시작%2 == 0 :
        시작 += 1
        continue
    print(시작)
    
    
    곱 = 1
    while 곱 < 끝 :
        print(f"{시작} * {곱} = {시작 * 곱}")    
        곱 +=1        
    
    시작 += 1
    
```

문제1)아래의 조건문에서 해당 음료수를 선택 했을 경우 금액을 알려주고 수량을 받는 임력창을 띄우세요.

```python
음료 = input("콜라, 사이다, 맥콜 중 하나의 음료수를 입력하세요.")
print(음료)

금액 = 0
if 음료 == '콜라':
    금액 = 500
if 음료 == '사이다':
    금액 = 450
if 음료 == '맥콜':
    금액 = 300    
    
print(f'{금액}원')

수량 = input("선택하신 음료의 수량을 입력하세요.")
수량 = int(수량)
print(수량)
```

문제2) 선택한 음료와 수량의 총합을 출력해주세요 예)콜라 3개 금액은 1500원입니다.

```python
print(f'{음료} {수량}개 금액은 {금액 * 수량}원 입니다.')
```

```python
돈 = input("선택 하신 음료를 구매하기 위해서 금액을 입력하세요.")
돈 = int(돈)
print(f'{돈}원를 받고 {돈 - (금액 * 수량)}원 거스름돈를 받으세요.')
```

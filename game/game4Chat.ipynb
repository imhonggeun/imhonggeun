```python
# 자판기에 등록된 음료와 가격
menu = {
    '콜라': 1500,
    '사이다': 1300,
    '커피': 1000,
    '물': 800
}

def show_menu():
    print("\n📋 자판기 메뉴:")
    for i, (item, price) in enumerate(menu.items(), 1):
        print(f"{i}. {item} - {price}원")

def get_choice():
    try:
        choice = int(input("\n번호를 선택하세요 (종료하려면 0): "))
        if choice == 0:
            return None
        if 1 <= choice <= len(menu):
            item = list(menu.keys())[choice - 1]
            return item
        else:
            print("잘못된 선택입니다.")
            return get_choice()
    except ValueError:
        print("숫자를 입력해주세요.")
        return get_choice()

def process_payment(item):
    price = menu[item]
    print(f"\n🧾 {item}의 가격은 {price}원입니다.")
    try:
        money = int(input("투입할 금액을 입력하세요: "))
        if money < price:
            print("금액이 부족합니다. 돈을 다시 돌려드립니다.")
        else:
            change = money - price
            print(f"{item}을(를) 구매하셨습니다. 거스름돈: {change}원")
    except ValueError:
        print("올바른 금액을 입력해주세요.")
        process_payment(item)

def run_vending_machine():
    print("🎉 자판기에 오신 것을 환영합니다!")
    while True:
        show_menu()
        choice = get_choice()
        if choice is None:
            print("자판기를 종료합니다. 감사합니다!")
            break
        process_payment(choice)

# 실행
run_vending_machine()
```

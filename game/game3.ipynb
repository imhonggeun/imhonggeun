문제 1) 1차원 배열(리스트)의 값의 평균값을 구하시오

```python
arr1 = [80,76,63,83,99,87,100,42,68,55]

def 평균값(list):
    합 = 0
    갯수 = 0
    for i in list :
        합 += i
        갯수 += 1
    return 합/ 갯수

print(평균값(arr1))
```

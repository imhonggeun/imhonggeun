1.자료형 - 숫자형(int)

```python
a = 1
b = 2
c = 3
print(a)
print(a-b)
print(a+c)
print(a/b)
print(a*b)
```

2.자료형 -문자열(str)

```python
a = '안녕하세요'
b = '반갑습니다.'
print(a)
print(a+b)
```

3.자료형-리스트

```python
list = [1,2,3,4,5]
print(type(list)) 
print(list[2])
list[2]='22'
print(list[2])
```

4.자료형-튜플 / 튜플형태는 변경이 불가능

```python
tuple = (1,2,3,4,5)
print(type(tuple))
print(tuple[0])
#tuple[0]='2'
print(tuple[0])
```

5.자료형-딕셔너리 KEY&VALUE 개념 

```python
dic = {1:'이름',2:'나이',3:'성별','4':'키'}
print(type(dic))
print(dic[1])
dic[1]='사람'
print(dic[1])
print(dic['4'])
```

6.자료형-불

```python
a = 1
b = 2
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

7.제어문 - if

```python
a = 1

if a == 2:
    print('성공')
else :
    print("실패")
```

8.제어문 - while

```python
a = 0
while a < 10:
    a = a+1
    print(f'while 안에 {a}')
    if a == 4:
        pass #break
print(f'while 밖에 {a}')
```

9.제어문 -for

```python
list = [10,20,30,40,50]

print(list)
for y in list:
    print(type(y))
```

```python
# 리스트 안에 튜블 안에 있는 튜플은 변경 불가능하고 튜플자체는 변경가능
a = [1,2,(3,4,5),6,7] # 리스트 안에 튜플이 있는 상황
a[0] =10 #리스트 안에 위치값
a[1] =20
#a[2] =30 #리스트 안에 있는 튜블자체(통으로)를 변경가능
a[3] =60
a[4] =70
print(a)

print(type(a[2]))
print(a[2][0]) # 리스트 안에 튜블안에 첫번째 자리 위치값 불러오기
a[2][0] = 30 # 리스트안에 튜블 안에 있는 첫번째 자리 값 변경
print(a) # 변경 불가능
```

```python
# 튜플 안에 있는 리스트 자체는 변경 불가능 /튜블안에 리스트 안에 있는 값은 변경 가능
b = (1,2,[3,4,5],6,7) #튜블 안에 리스트
#b[0] = 10
#b[1] = 20
b[2] = 30 # 튜블 안에 있는 리스트 자체는 변경 불가능
#b[3] = 40
#b[4] = 50
print(b)
```

```python
# 튜블 안에 리스트 안에 값 변경 가능
b = (1,2,[3,4,5],6,7)
print(b[2])
print(type(b[2])) # 튜블 안에 있는 리스트 

print([2][0]) # 튜블에서 3번째(2) 에서 0번(1) 자리값 불러오기
b[2][0] = 30 # 튜블 안에 있는 리스트에서 첫번째 값을 변경
print(b[2][0]) # 튜블 안에 있는 리스트 값은 변경 가능
print(b)
```
